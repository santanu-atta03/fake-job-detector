# 01. Exploratory Data Analysis (EDA) - Fake Job Postings

This notebook performs comprehensive Exploratory Data Analysis (EDA) on the `fake_job_postings.csv` dataset (~17,880 rows).

### Key Objectives:
1. **Class Distribution Analysis**: Quantify the severe class imbalance (~4.8% fraudulent job postings vs 95.2% legitimate).
2. **Metadata Completeness**: Analyze missing company profiles, company logos, screening questions, and salary information.
3. **Text Statistics**: Distribution of job description length, word counts, and uppercase ratio across classes.
4. **Scam Indicators**: Evaluate frequency of urgency words, money language, scam keywords, and free email domains.

```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Add ml/src to Python path
sys.path.append(str(Path.cwd().parent / "src"))
from data_loader import load_raw_data, clean_job_data
from feature_engineering import create_scam_features

# Matplotlib style configuration
plt.style.use('ggplot')
plt.rcParams['figure.figsize'] = (10, 5)
plt.rcParams['font.size'] = 11
print("Libraries loaded successfully.")
```

## 1. Load and Inspect Raw Dataset

```python
raw_df = load_raw_data()
cleaned_df = clean_job_data(raw_df)

print("Dataset Shape:", cleaned_df.shape)
display(cleaned_df.head(3))
display(cleaned_df.info())
```

## 2. Target Variable & Class Imbalance Analysis

```python
target_counts = cleaned_df['fraudulent'].value_counts()
target_props = cleaned_df['fraudulent'].value_counts(normalize=True) * 100

summary_df = pd.DataFrame({'Count': target_counts, 'Percentage (%)': target_props.round(2)})
summary_df.index = ['Legitimate (0)', 'Fraudulent (1)']
print(summary_df)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.barplot(x=summary_df.index, y=summary_df['Count'], ax=axes[0], palette=['#2ecc71', '#e74c3c'])
axes[0].set_title('Class Count Distribution')
axes[0].set_ylabel('Number of Postings')

axes[1].pie(summary_df['Count'], labels=summary_df.index, autopct='%1.2f%%', colors=['#2ecc71', '#e74c3c'], startangle=140)
axes[1].set_title('Class Proportion Pie Chart')
plt.tight_layout()
plt.show()
```

## 3. Metadata Completeness & Missing Company Information

```python
# Compare presence of logo, company profile, and screening questions by fraud label
meta_cols = ['has_company_logo', 'has_questions']
meta_df = cleaned_df.groupby('fraudulent')[meta_cols].mean() * 100
meta_df.index = ['Legitimate', 'Fraudulent']

# Add company profile presence
cleaned_df['has_company_profile'] = (cleaned_df['company_profile'].str.len() > 0).astype(int)
meta_df['has_company_profile'] = cleaned_df.groupby('fraudulent')['has_company_profile'].mean() * 100

print("Percentage of postings with metadata:")
    display(meta_df.round(2))

meta_df.T.plot(kind='bar', figsize=(10, 5), color=['#2ecc71', '#e74c3c'])
plt.title('Presence of Key Metadata Features (% of postings)')
plt.ylabel('Percentage (%)')
plt.xticks(rotation=0)
plt.show()
```

## 4. Text Length and Word Count Distribution

```python
cleaned_df['text_length'] = cleaned_df['job_text'].str.len()
cleaned_df['word_count'] = cleaned_df['job_text'].str.split().str.len()

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.histplot(data=cleaned_df, x='word_count', hue='fraudulent', bins=50, kde=True, ax=axes[0], palette=['#2ecc71', '#e74c3c'])
axes[0].set_title('Word Count Distribution by Class')
axes[0].set_xlim(0, 1500)

sns.boxplot(data=cleaned_df, x='fraudulent', y='text_length', ax=axes[1], palette=['#2ecc71', '#e74c3c'])
axes[1].set_title('Text Length Boxplot by Class')
axes[1].set_xticklabels(['Legitimate', 'Fraudulent'])
axes[1].set_ylim(0, 10000)

plt.tight_layout()
plt.show()
```

## 5. Domain Scam Feature Correlation Analysis

```python
scam_df = create_scam_features(cleaned_df)
scam_df['fraudulent'] = cleaned_df['fraudulent']

correlations = scam_df.corr()['fraudulent'].sort_values(ascending=False)
print("Top Positive Correlations with Fraudulent Label:")
print(correlations.head(10))
print("\nTop Negative Correlations with Fraudulent Label:")
print(correlations.tail(10))

plt.figure(figsize=(10, 8))
top_features = correlations.drop('fraudulent').abs().sort_values(ascending=False).head(15).index
sns.heatmap(scam_df[top_features.union(['fraudulent'])].corr(), annot=True, fmt='.2f', cmap='coolwarm', vmin=-0.5, vmax=0.5)
plt.title('Correlation Heatmap of Top Scam Features & Target')
plt.tight_layout()
plt.show()
```

## 6. Key EDA Findings & Insights

1. **Severe Imbalance**: Fraudulent jobs constitute only **~4.84%** (866 instances) of the entire dataset. Stratified sampling and cost-sensitive/balanced models (e.g. `class_weight='balanced'`) are mandatory.
2. **Missing Metadata as Fraud Signals**: Fraudulent postings are significantly less likely to have a company profile or company logo present.
3. **Text Length**: Legitimate postings tend to be longer and more comprehensive than scam listings.
4. **Domain Keywords**: Excessive urgency, suspicious payment language, and free email domains strongly correlate with scam listings.
