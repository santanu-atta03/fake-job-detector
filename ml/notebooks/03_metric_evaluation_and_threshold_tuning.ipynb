# 03. Metric Evaluation & Threshold Tuning

This notebook evaluates the final production model on the held-out test dataset (`3,576` samples) and performs decision threshold optimization.

### Key Notebook Highlights:
1. **Test Set Metric Evaluation**: ROC-AUC, PR-AUC, Precision, Recall, F1-Score.
2. **Confusion Matrix Analysis**: Visualizing True Positives, False Positives, True Negatives, False Negatives.
3. **Precision-Recall Trade-off**: Plotting precision/recall curves over probability thresholds.
4. **SHAP Feature Importance**: Explaining top features influencing fraud predictions.

```python
import sys
from pathlib import Path
    import joblib
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import (
    roc_curve, precision_recall_curve, roc_auc_score, average_precision_score,
    confusion_matrix, classification_report, precision_score, recall_score, f1_score
)

sys.path.append(str(Path.cwd().parent / "src"))
from data_loader import PROCESSED_DATA_DIR
from feature_engineering import create_scam_features, combine_features

plt.style.use('ggplot')
print("Modules successfully loaded.")
```

## 1. Load Test Dataset & Trained Production Artifacts

```python
BASE_DIR = Path.cwd().parent.parent
MODEL_DIR = BASE_DIR / "ml" / "models"

test_path = PROCESSED_DATA_DIR / "test.csv"
test_df = pd.read_csv(test_path)

model = joblib.load(MODEL_DIR / "model.pkl")
tfidf = joblib.load(MODEL_DIR / "tfidf.pkl")
threshold = joblib.load(MODEL_DIR / "threshold.pkl")
feature_names = joblib.load(MODEL_DIR / "feature_names.pkl")

print(f"Loaded Test Set: {test_df.shape}")
print(f"Current Production Decision Threshold: {threshold}")
```

## 2. Generate Model Probabilities & Plot ROC / PR Curves

```python
scam_features = create_scam_features(test_df)
text_vector = tfidf.transform(test_df["job_text"].fillna(""))
X_test = combine_features(text_vector, scam_features)
y_test = test_df["fraudulent"].values

y_probs = model.predict_proba(X_test)[:, 1]

fpr, tpr, _ = roc_curve(y_test, y_probs)
precision_arr, recall_arr, _ = precision_recall_curve(y_test, y_probs)
roc_auc = roc_auc_score(y_test, y_probs)
pr_auc = average_precision_score(y_test, y_probs)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(fpr, tpr, color='#3498db', lw=2, label=f'ROC Curve (AUC = {roc_auc:.4f})')
axes[0].plot([0, 1], [0, 1], color='gray', linestyle='--')
axes[0].set_title('Receiver Operating Characteristic (ROC) Curve')
axes[0].set_xlabel('False Positive Rate')
axes[0].set_ylabel('True Positive Rate')
axes[0].legend()

axes[1].plot(recall_arr, precision_arr, color='#9b59b6', lw=2, label=f'PR Curve (AUC = {pr_auc:.4f})')
axes[1].set_title('Precision-Recall Curve')
axes[1].set_xlabel('Recall')
axes[1].set_ylabel('Precision')
axes[1].legend()

plt.tight_layout()
plt.show()
```

## 3. Threshold Sweep & Precision-Recall Trade-Off

```python
thresholds = np.linspace(0.05, 0.95, 91)
precs, recs, f1s = [], [], []

for t in thresholds:
    preds = (y_probs >= t).astype(int)
    precs.append(precision_score(y_test, preds, zero_division=0))
    recs.append(recall_score(y_test, preds, zero_division=0))
    f1s.append(f1_score(y_test, preds, zero_division=0))

plt.figure(figsize=(10, 5))
plt.plot(thresholds, precs, label='Precision', color='#2ecc71', lw=2)
plt.plot(thresholds, recs, label='Recall', color='#e74c3c', lw=2)
plt.plot(thresholds, f1s, label='F1-Score', color='#f39c12', lw=2)
plt.axvline(x=threshold, color='black', linestyle='--', label=f'Current Threshold ({threshold})')
plt.title('Metrics vs Decision Threshold')
plt.xlabel('Probability Threshold')
plt.ylabel('Score')
plt.legend()
plt.show()
```

## 4. Confusion Matrix Visualization at Current Threshold

```python
y_preds = (y_probs >= threshold).astype(int)
cm = confusion_matrix(y_test, y_preds)

plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Legitimate', 'Fraudulent'], yticklabels=['Legitimate', 'Fraudulent'])
plt.title(f'Confusion Matrix (Threshold = {threshold})')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.show()

print("Classification Report:")
print(classification_report(y_test, y_preds, target_names=['Legitimate', 'Fraudulent']))
```

## 5. SHAP Feature Importance Analysis

```python
try:
    import shap
    explainer = shap.TreeExplainer(model)
    # Sample 100 test instances for fast SHAP summary
    sample_idx = np.random.choice(X_test.shape[0], size=min(200, X_test.shape[0]), replace=False)
    X_sample = X_test[sample_idx].toarray()
    shap_vals = explainer.shap_values(X_sample)
    
    plt.figure(figsize=(10, 6))
    if isinstance(shap_vals, list):
        shap_vals_target = shap_vals[1]
    else:
        shap_vals_target = shap_vals
        
    shap.summary_plot(shap_vals_target, X_sample, feature_names=feature_names, max_display=15)
except Exception as e:
    print("SHAP analysis notice:", e)
```
